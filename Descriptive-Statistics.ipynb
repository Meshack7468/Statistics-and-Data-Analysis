# Measures of Central Tendency
## Mean

* Mean is the average value of a dataset, it acts as a primary measure to show the center point of a data distribution.
* Very high numbers(outliers) can pull the mean up while very low numbers(outliers) can pull the mean down and skew the average.
* Types of mean we have arithmetic mean, weighted mean, and geometric mean 

Formular:
$$ \bar{x} = \frac{\sum x}{n} $$

Where:
* x̄ represents the mean (average) of the data set.
* $\sum$ (Sigma) represents the **"sum of"** or to add everything up
* x represents each individual data value in the dataset.
* n = Total number of values in the sample

```python
# Calculate the mean for the numeric values
mean_values = df[numerical_columns].mean()

print("Mean:")
print(mean_values)
```

* The mean values for Age, Experience_Years, and Salary are 35.46 years, 12.37 years, and 115,381.50, respectively.
* The mean gives us an overall measure of the center of the age, experience, and salary of the employees. However, mean can be affected by unusually high or low values (outliers).
For this reason, the median is often preferred as a more stable measure of central tendency.

## Median
* The median is the middle value when the data is arranged from smallest to largest or vice versa.

```python
median_values = df[numerical_columns].median()

print("Median:")
print(median_values)
```

* Median Age (34 years): Half of the employees are 34 years old or younger, while the other half are 34 years old or older
* Median Experience (10 years): Half of the employees have 10 years of experience or less, while the remaining half have 10 years or more.
* Median Salary (120,000): Half of the employees earn 120,000 or less, while the other half earn 120,000 or more.
* Robustness to Outliers: Unlike the mean, the median is highly resistant to unusually high or low values (outliers).

## Mode
* The mode is the value that appears most

```python
mode_values = df[numerical_columns].mode()

print("Mode:")
print(mode_values)
```

* The most frequently occurring Employee age is 30 years.
* The most frequently occurring Employee experience level is 1 year.
* The most frequently occurring salary is 70,000.

# Measures of Dispersion

## Range
* The range measures the difference between the **maximum** and **minimum** values.

```python
range_values = df[numerical_columns].max() - df[numerical_columns].min()

print("Range:")
print(range_values)
```

* Age Range (39 years): The difference between the youngest and oldest employee is 39 years.
* Experience Range (37 years): The gap between the employee with the least experience and the one with the most experience is 37 years.
* Salary Range (190,000): The range between the lowest and highest annual salary is 190,000.
* A larger range indicates a greater variation between the minimum and maximum data points in the workforce.
*  Because the range relies solely on two extreme values, it fails to capture how the rest of the data is distributed. To gain an in-depth understanding of data dispersion, metrics such as standard deviation, variance, and the Interquartile Range (IQR) are critical.


## Variance
* Variance is a statistical measurement that shows the dispersion of a dataset from the mean (average).

* Sample Variance (s²): Use this when the data is only a small group (a sample) taken from a larger total population.

formular:
$$ s^2 = \frac{\sum (x - \bar{x})^2}{n - 1} $$

where:

* **$\sum$** = Summation (all the calculated values)

* **$\bar{x}$** = x-bar (the Mean/average of the data set)


* Population Variance (σ²): Used when your data includes the entire group studied.

formular:
$$\sigma^2 = \frac{\sum (x_i - \mu)^2}{N}$$

Where:
* **$s^2$** = Sample Variance
*   $\sigma^2$ = Population variance
*   $\sum$ = Summation (add up all the results)
*   **$x$** = Individual data points in the dataset
*   $x_i$ = Each individual data point
*   $\mu$ = Population mean (average)
*   $N$ =  Total number of values in the population
*   **$n$** = Total number of values in the sample

```python
variance_values = df[numerical_columns].var()

print("Variance:")
print(variance_values)
```

* Age: The variance is 100.00 years. This indicates how much employee ages vary around the mean age.
* Experience: The variance is 83.70 years, showing the variation in years of experience around the mean.

## Standard Deviation
* Standard deviation shows how much the values differ from the mean.
* Here we also have population and sample std dev

* Population std is used when studying the entire group(population)

$$\sigma = \sqrt{\frac{\sum (x_i - \mu)^2}{N}}$$

* Sample std dev used when studying a sample
$$ s = \sqrt{\frac{\sum (x - \bar{x})^2}{n - 1}} $$

where: 
*   $\sigma$ = Population standard deviation
*   $s$ = Sample standard deviation
*   **$\sqrt{\dots}$** = Square Root 
*   $\mu$ = Population mean
*   **$x$** = Individual data points in the data set
*   $\bar{x}$ = Sample mean
*   $N$ = Total number of values in the population
*   $n$ = Total number of values in the sample

```python
std_values = df[numerical_columns].std()

print("Standard Deviation:")
print(std_values)
```

The standard deviation tells us how much the values differ from the mean. A larger standard deviation indicates greater variability in the data.

* Age std is 10.00 years
* The average age is 35.46 years.
* Employee ages vary by around 10 years from the average.


* Experience_Years std is 9.15 years
* The average experience is 12.37 years.
* Experience levels vary by  9.15 years from the average.

* Salary std is 46,066.14
* The average salary is 115,381.50.
* Salaries vary by about 46,066 from the average salary.


## Interquartile Range (IQR)

```python
Q1 = df[numerical_columns].quantile(0.25)
Q3 = df[numerical_columns].quantile(0.75)

IQR = Q3 - Q1

print("Interquartile Range:")
print(IQR)
```

* Age: The IQR is 16 years, suggesting the middle 50% of employees ages are spread across a 16-year range.
* Experience: The IQR is 14 years, suggesting the middle 50% of employees experience levels are spread across a 14-year range.
* Salary: The IQR is 80,000, meaning the middle 50% of employee salaries are spread across an 80,000 range.
* Unlike the range, the IQR focuses on the middle 50% of observations and is less affected by extreme values.
* IQR is particularly useful when identifying potential outliers using the lower and upper bounds.

# Visualizations

## Histograms

* A histogram is a graphical graph that shows how **numerical data** is distributed 

```python
for column in numerical_columns:
    plt.figure(figsize=(8, 5))

    sns.histplot(data=df, x=column, kde=True)

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.show()
```

* Age: Most employees are between 25 and 30 years old, with the number of employees generally decreasing as age increases.
* Years of Experience: Most employees have between 0 and 10 years of work experience, with the number of employees gradually declining as years of experience increase.
* Salary: Employee salaries are mainly concentrated around 75,000 and within the 125,000–150,000 range, indicating higher representation within these salary levels.

## Box Plots

* A Box Plot is a visual graph that shows the distribution of a numerical data.
* It also helps us spot possible outliers.
* It shows how data is spread out using five key numbers. 

![Box Plot](https://www.simplypsychology.org/wp-content/uploads/image-3.jpeg)

Main Parts of the Graph
* The Box: Covers the middle 50% of the data (from Q1 to Q3).
*  Its length is called the Interquartile Range (IQR).
*  The Whiskers: Lines that stretch out from the box to the highest and lowest normal values.
*  Outliers: Single dots outside the whiskers that show unusual or extreme data points.

* Minimum: The smallest value in the data set, shown at the end of the bottom or left whisker.
* Lower Quartile (Q1): The 25th percentile, which marks the start of the main box.
* Median (Q2): The middle number of the data set, shown as a line inside the box.
* Upper Quartile (Q3): The 75th percentile, which marks the end of the main box.
* Maximum: The largest value in the data set, shown at the end of the top or right whisker.

```python
for column in numerical_columns:
    plt.figure(figsize=(8, 4))

    sns.boxplot(data=df, x=column)

    plt.title(f"Box Plot of {column}")
    plt.xlabel(column)

    plt.show()
```

We Use Box Plots For 
* Quick comparisons: The box plots allow us to compare employee age, experience, and salary side by side. For example, the median age is 34 years, median experience is 10 years, and median salary is 120,000. The lower quartile (Q1) is 27 years for age, 5 years for experience, and 70,000 for salary. The upper quartile (Q3) is 43 years for age, 19 years for experience, and 150,000 for salary.
* Spotting skewness: The box plots show that employee age, experience, are slightly right-skewed. This means most values cluster toward the lower end of the data
* Box plot also help in spotting outliers: 

## Bar Charts

* Bar charts are useful for showing categories

```python
for column in categorical_columns:
    plt.figure(figsize=(8, 5))

    sns.countplot(data=df, x=column)

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Number of Employees")
    plt.xticks(rotation=45)

    plt.show()
```

Observations
* Gender: The gender distribution is relatively balanced, with no major difference between the categories.
* Department: Employees are relatively evenly distributed across the available departments, with no department showing imbalance.
* Job Title: The most common job titles are Manager, Analyst, Executive, Engineer, and Intern, respectively.
* Education Level: Most employees hold a Master's degree, followed by Bachelor's and PhD qualifications, respectively.
* Location: The distribution of employees across the different locations is relatively even.

## Scatter Plot

* This scatter plot shows the relationship between years of work experience and salary

* So there is a positive correlation between experience and salary
