* Statistics is the process of collecting, organizing, analyzing, interpreting, and communicating data.

It helps us understand data, identify patterns, and make informed conclusions.

## 2. Data and Variables

A variable is a characteristic that can have different values.

For example:

Age can have different values for different employees.
Salary can vary between employees.
Gender can have different categories.

Data is the information collected for a study or analysis.

## Types of Data
## 3. Qualitative & Quantitative Data

* Qualitative data describes qualities, attributes, or labels using words, images, or observations. You cannot do math with these values.

Examples include:

Gender(Male, Female),
Employee department,
Education Level

* Quantitative data represents **numerical values** that can be measured or counted.

Examples include:

Age,
salary,
Experience Years
### Categorical Data

* Categorical data is a specific type of qualitative data that sorts observations into distinct groups or categories.

For example, Department can contain categories such as Finance, HR, and Engineering.

### Binary Data

* Binary data is a type of categorical data with only two possible categories.

Examples include:

* Yes / No
* True / False
*  0 / 1

### Multiclass Data
* Multiclass data is categorical data that has three or more distinct outcomes.
  
### Ordinal Data

* Ordinal data is categorical data with a meaningful order or ranking.

Examples include:

* Low / Medium / High
* Poor / Good / Excellent

### Nominal Data
* Category dat with no specific order (e.g., eye color, red/blue, male/female).

## 4. Discrete & Continuous Data
### Discrete Data

* Discrete data consists of countable values, usually whole numbers.

Examples include:

* Number of students in a class
* Number of goals scored in a match
* Number of employees in a department
### Continuous Data

* Continuous data can take any numerical value within a range, including decimals and fractions.

Examples include:

* Height
* Weight
* Temperature

For example, a person's height could be 175.5 cm.


## 5. Population and Sample

* A population is the entire group being studied.

i.e, if a company has 10,000 employees, all 10,000 employees make up the population.

* A sample is a smaller group selected from the population.

i.e if we select 200 employees from the 10,000 employees, those 200 employees form the sample.

We use samples because studying an entire population may be time-consuming, expensive, or difficult. A carefully selected sample can provide useful information about the larger population.



## 6. Parameter & Statistic

* A parameter is a numerical value that describes an entire population.

For example, the average salary of all 10,000 employees is a population parameter.

* A statistic is a numerical value that describes a sample.

For example, the average salary of 200 selected employees is a sample statistic.

## 7 Descriptive & Inferential Statistics
### Descriptive Statistics

Descriptive statistics describes, organizes, and summarizes the data we have.

This includes:

* Measures of Central Tendency: Describe the center or typical value of the data, such as mean, median, and mode.
* Measures of Variability: Describe how spread out the data is, such as range, variance, standard deviation, and interquartile range (IQR).
* Distribution Measures: Help describe the distribution of the data, such as percentiles, skewness, and kurtosis.
* Visual Tools: Help summarize data using histograms, bar charts, scatterplots, and boxplots.
### Inferential Statistics

Inferential statistics uses data from a sample to make conclusions or estimates about a larger population.

This includes:

* Hypothesis Testing: Uses sample data to test claims about a population.
* Confidence Intervals: Provide a range of values used to estimate a population parameter.
* Correlation: Measures the strength and direction of the relationship between two numerical variables.
* Regression: Models the relationship between variables and can be used to make predictions.
