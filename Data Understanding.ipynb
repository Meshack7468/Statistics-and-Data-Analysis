```python
# Classify Variables
# Separate numerical and categorical variables

categorical_columns = df.select_dtypes(include="object").columns
numerical_columns = df.select_dtypes(include=np.number).columns

print("Categorical Variables:")
print(categorical_columns)

print("\nNumerical Variables:")
print(numerical_columns)
```

```python
# Explore Numerical Variables
df.describe()
```

* Age: Employees are between 21 and 60 years old, with an average age of about 35 years. Half of the employees are aged between 27 and 43 years, suggesting the dataset is largely composed of young to middle-aged employees.
* Experience Years: Employees have between 0 and 37 years of experience, with an average 12 years. The median experience is 10 years, while 50% of employees have between 5 and 19 years of experience.
* Salary: Salaries range from 25,000 to 215,000, with an average/mean salary of  115,382. The median(50%) salary is 120,000, while half of the employees earn between 70,000 and 150,000.

```python
# Examine Data Distribution for Numerical variables

numerical_columns = numerical_columns.drop("Employee_ID")

for column in numerical_columns:
    plt.figure(figsize=(8, 5))
    sns.histplot(data=df, x=column, kde=True)
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.show()
```

```python
## For Categorical variables

categorical_columns = categorical_columns.drop("Name", errors="ignore")

for column in categorical_columns:
    plt.figure(figsize=(8, 5))
    sns.countplot(data=df, x=column)
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.show()
```

* Gender: The gender distribution is relatively balanced, with no major difference between the categories.
* Department: Employees are relatively evenly distributed across the available departments, with no department showing a substantial imbalance.
* Job Title: The most common job titles are Manager, Analyst, Executive, Engineer, and Intern, respectively.
* Education Level: Most employees hold a Master's degree, followed by Bachelor's and PhD qualifications, respectively.
* Location: The distribution of employees across the different locations is relatively even.
